```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import Pipeline

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name, sep=None, engine='python')

df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
df.describe()
```

```python
print(df['quality'].value_counts().sort_index())

df['quality'].value_counts().sort_index().plot(
    kind='bar',
    xlabel='Wine Quality',
    ylabel='Count',
    title='Distribution of Wine Quality'
)

plt.show()
```

```python
X = df.drop('quality', axis=1)
y = df['quality']

print("Input Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
ann = MLPRegressor(
    hidden_layer_sizes=(64, 32),
    activation='relu',
    solver='adam',
    learning_rate_init=0.001,
    max_iter=500,
    early_stopping=True,
    validation_fraction=0.1,
    n_iter_no_change=20,
    random_state=42
)
```

```python
model = Pipeline([
    ('scaler', StandardScaler()),
    ('ann', ann)
])
```

```python
model.fit(X_train, y_train)

print("ANN Training Completed")
```

```python
y_pred = model.predict(X_test)

print("Actual Quality:")
print(y_test.values[:10])

print("\nPredicted Quality:")
print(np.round(y_pred[:10], 2))
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error:", round(mae, 4))
print("Mean Squared Error:", round(mse, 4))
print("Root Mean Squared Error:", round(rmse, 4))
print("R2 Score:", round(r2, 4))
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred, alpha=0.5)

plt.xlabel("Actual Wine Quality")
plt.ylabel("Predicted Wine Quality")
plt.title("Actual vs Predicted Wine Quality")

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle='--'
)

plt.show()
```

```python
loss_values = model.named_steps['ann'].loss_curve_

plt.figure(figsize=(8, 5))
plt.plot(loss_values)

plt.xlabel("Training Iteration")
plt.ylabel("Loss")
plt.title("ANN Training Loss Curve")

plt.show()
```

```python
results = pd.DataFrame({
    'Actual Quality': y_test.values,
    'Predicted Quality': np.round(y_pred, 2)
})

results['Absolute Error'] = np.abs(
    results['Actual Quality'] - results['Predicted Quality']
)

results.head(20)
```
